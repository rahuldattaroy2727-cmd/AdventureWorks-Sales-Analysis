```python
import pandas as pd
import matplotlib.pyplot as plt


rev_by_region  = pd.read_csv('/content/Revenue_by_region.csv')
```

 ***REVENUE BY TERRITORY***



```python
revbar =rev_by_region[['region','revenue']]
```

```python
revbar
```

```python

plt.figure(figsize =(6,6),facecolor ='grey',)

def format_revenue(value):
  if value >= 1_000_000:
    return f'{round((value/1_000_000),2)}M'
  elif value >= 1_000:
      return f'{round((value/1_000),2)}K'
  else:
      return f'{value}'
barhs =plt.barh(revbar['region'],revbar['revenue'],color ='green',edgecolor='blue')

for barh in barhs:
  plt.text(barh.get_width()+0.4,barh.get_y()+(barh.get_height())/2,format_revenue(barh.get_width()),ha = 'left',va='center')
  plt.xlim(0, revbar['revenue'].max() * 1.15)
plt.ylabel('Regions',color ='white')
plt.xlabel('Revenue(M)',color = 'white')
plt.yticks(color = 'white')
plt.xticks(color = 'white')
plt.grid(alpha = 0.2,color = 'grey')
plt.title('Territory Revenue',color = 'white')
##############################################
#Insights:
#Australia generated the highest revenue, while Central,
# Northeast, and Southeast contributed very little compared with the leading territories.
###############################################
```

# **REVENUE BY YEAR**

```python
by_year = pd.read_csv('/content/revenue_by_year.csv')
```

```python
by_year[['years','revenue']]
```

```python
by_year['years']=by_year['years'].astype(int)
```

```python
from matplotlib.lines import lineStyles
plt.figure(figsize=(6,6),facecolor ='grey')
plt.plot(by_year['years'],by_year['revenue'],color ='green',marker = 'o',linestyle = 'dashed',markerfacecolor='blue')
plt.ylim(5.0,by_year['revenue'].max()*2)
plt.xticks(by_year['years'].astype(int),rotation = 45,color ='white')
plt.yticks(color='white')
plt.xlabel('Years',color ='white')
plt.ylabel('Revenue',color='white')
plt.gca().yaxis.set_major_formatter(lambda x ,pos:f'{x/1_000_000:g}M')
plt.grid(alpha = 0.4)
plt.title('Yearly Revenue',color= 'white')
##################################################
#Insights:
#Revenue increased significantly from 2020 to 2021, followed by a slight decline in 2022.
###################################################
```

TOP 10 PRODUCTS BY **REVENUE**

```python
top_10 = pd.read_csv('/content/Top_10_products_by_revenue.csv')
```

```python
top_10
```

```python
def format_revenue(value):
  if value >= 1_000_000:
    return f'{round((value/1_000_000),2)}M'
  elif value >= 1_000:
      return f'{round((value/1_000),2)}K'
  else:
      return f'{value}'

plt.figure(figsize = (6,6),facecolor='grey')
barhs = plt.barh(top_10['productname'],top_10['product_revenue'],color='green',edgecolor='blue')

plt.ylabel('products',color='white')
plt.xlabel('revenue',color='white')
plt.xlim(0.0,top_10['product_revenue'].max()*1.2)
plt.xticks(color='white')
plt.yticks(color='white')
for barh in barhs:
  plt.text(barh.get_width()+0.8,barh.get_y()+(barh.get_height())/2,format_revenue(barh.get_width()),color ='blue')
plt.grid(alpha =0.2,color='grey')
plt.title('Top10 Products by Revenue')
###############################################################
#Insights:
#Revenue was concentrated among a small group of top-performing products,
#with the highest-ranked products generating significantly more revenue than the lower-ranked products in the top 10.
###############################################################
```
